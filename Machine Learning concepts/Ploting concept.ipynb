```python
# NORMAL DISTRIBUTION
#Example 1
#The annual salaries of a employee in a large company approximately ND is 50000 and std is 10000 
#What percent of people earn less than $60000

from scipy.stats import norm
nd=norm(50000,10000)
x1=nd.cdf(60000)
x1
```

```python
#Example 2
#a radar units is used to measure speeds cars on highway. the speed are normally distributed with 90 kmh and 
# std is 10 probability that a car picked at random is more than 100 kmh 
nd=norm(90,10)
x1=1-nd.cdf(100)
x1
```

```python
#Example 3 
# for certain type of mobile the length of the time betwee charge of the battery is normally distributed 10 hrs and std is 20 mnts
# John owns of these mobiles want to know probability of that the length of time will be between 10 and 11 hrs.
nd=norm(600,20)
x1=nd.cdf(660)-nd.cdf(600)
x1
```
