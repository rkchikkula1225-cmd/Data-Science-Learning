```python
import nltk
nltk.download("all")
```

```python
#sample text
text="""OMG!!! I bought this Samsung Galaxy S22 Ultra  last week from Amazon... Seriously, it's A-W-E-S-O-M-E!!! 

The camera quality is sooo good (better than my last iPhone 12)! Battery life? Not bad, but could be better lol... 

Also, delivery  was super fast!! But hey, price-wise it's kinda high ... Overall, I'm happy with it, totally worth it. 

#Samsung #GalaxyS22 #mobilepurchase Can't wait to recommend this to my friends & fam! 

Btw, shoutout to @Amazon support team, they were helpful when I had payment issues! 10/10! """
print("original data:")
print(text)
```

```python
#lower casing
text=text.lower()
print("lower case text:")
print(text)
```

```python
import re

# remove URLs
text = re.sub(r'http\S+|www\S+|https\S+', ' ', text)

# remove mentions and hashtags
text = re.sub(r'@\w+|#\w+', ' ', text)

# remove special characters
text = re.sub(r'[^\w\s]', ' ', text)

# remove extra spaces
text = re.sub(r'\s+', ' ', text).strip()

print("After text cleaning:")
print(text)
```

```python
#Tokenization
tokens=nltk.word_tokenize(text)
print("after tokenization:")
print(tokens)
print(len(tokens))
```

```python
#stopword words remove
from nltk.corpus import stopwords
stopwords_list=stopwords.words('english')
print("total words:", len(stopwords_list))
print("list of words:", stopwords_list)
stopwords_list.append(101)
```

```python
#
filtered_tokens=[]
for i in tokens:
    if i not in stopwords_list:
        filtered_tokens.append(i)
print("Total words:", len(filtered_tokens))
print(filtered_tokens)
```

```python
#stemming
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
stemmed_tokens=[stemmer.stem(i) for i in filtered_tokens]
print(stemmed_tokens)
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
lemmatized_tokens=[lemmatizer.lemmatize(i) for i in filtered_tokens]
print(lemmatized_tokens)
```

```python
#join the tokens back from a cleaned sentence
cleaned_text=' '.join(lemmatized_tokens)
print("\n cleaned text after full preprocessing:")
print(cleaned_text)
```
