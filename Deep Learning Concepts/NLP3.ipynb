```python
from gensim.models import Word2Vec
```

```python
from nltk.tokenize import word_tokenize
import nltk
```

```python
sentences = [

    "NLP helps computers understand and interpret human language",

    "NLP tasks include text classification, machine translation, and sentiment analysis",

    "NLP models are trained on large datasets of text to learn patterns and relationships.",

    "NLP applications are used in various fields, such as customer service, healthcare, and marketing."

]
```

```python
#Define the word 2Vec model using skigram
model=Word2Vec(
    Sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=0,
    compute_loss=True)
```

```python
#Train model
model.train(
    tokenized_sentences,
    total_examples=len(tokenized_sentences)
    epochs=100)
similar_words=model.wv.most_similar('nlp', topn=5)
similar_words
print("top 5 words similar to 'nlp':")
for similar_word, score\nn similar_words:
    print(f"{similar_word}:{score:.4f}")
```

```python
#find the nearest word
nearest_word=similar_words[0][0]
print(f"\n Nearest words to 'nlp': {nearest_word}")
```
