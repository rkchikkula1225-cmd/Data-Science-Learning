```python
import pandas as pd
import numpy as np
sample=pd.read_csv('D:\csv files\Bulbs_Lifespan.csv')
```

```python
# ONE SAMPLE T_TEST

import pandas as pd
from scipy.stats import ttest_1samp
pop_mean=1000                                     # Hypothesized population mean
t_statistic, p_value=ttest_1samp(sample,pop_mean) # perform the one sample t-test
alpha=0.05                                        # set alpha value = 0.05(5% risk of type 1 error)
print("T_statistics:", np.round(t_statistic,4))   # to round 4 four decimals
print("P_value:  ", np.round(p_value,4))
if p_value< alpha:                                 # to check the weather p value less with alpha value 0.05
    print("Reject the null the hypothesis: The mean lifespan is significantly different from the 1000 hrs.")
else:
    print("Fail to Reject the null the hypothesis: no significant different from the 1000 hrs.")
                   
```

```python
# for alternative hypothesis: less than (H1: μ<1000
# this mean we are perfroming the left tailed t-test
if t_statistic<0:
    one_tailed_p_value=p_value/2
    if one_tailed_p_value < alpha:
        print("Reject the null the hypothesis: The mean lifespan is significantly different from the 1000 hrs.")
    else:
        print("Fail to Reject the null the hypothesis: no significant different from the 1000 hrs.")
        
```

```python
# for alternative hypothesis: greater than (H1: μ>1000
# this mean we are perfroming the Right tailed t-test
if t_statistic>0:
    one_tailed_p_value=p_value/2
    if one_tailed_p_value < alpha:
        print("Reject the null the hypothesis: The mean lifespan is significantly different from the 1000 hrs.")
    else:
        print("Fail to Reject the null the hypothesis: no significant different from the 1000 hrs.")  
else:
    print("Fail to Reject the null the hypothesis: test statistic does not support H1 (mean>1000).")  
```

```python
# ONE SAMPLE Z_TEST
import pandas as pd
import numpy as np
df=pd.read_csv('D:\csv files\Bulbs_Lifespan.csv')
```

```python
# ONE SAMPLE Z_TEST
from statsmodels.stats.weightstats import ztest     #import z_test function from stats models                                                 # this function will perform the one sample ztest
population_mean=1000                                # Given Data
alpha=0.05                                          #set alpha value = 0.05(5% risk of type 1 error)

z_stat, p_value=ztest(df["Lifespan (Hours)"], value= population_mean, alternative='two-sided')  #step 1 to perform the ztest
if p_value<alpha:                                                                               # step 2 Decision
    
    conclusion="Reject the null the hypothesis: The avg lifespan is significantly different from the 1000 hrs."
else:
    conclusion="Fail to Reject the null the hypothesis: there is no significant different from the 1000 hrs."
print(f"z_statistic:{z_stat:.2f}")
print(f"p_value:{p_value:.4f}")
print(f"conclusion:{conclusion}")
```

```python
import pandas as pd
import numpy as np
df=pd.read_csv('D:\csv files\ecommerce_two_sample_ztest_dataset.csv')
```

```python
from statsmodels.stats.weightstats import ztest 
email_group=df[df['Group']=='Email Marketing'] ['Purchase_Value']
social_group=df[df['Group']=='Social Media Ads'] ['Purchase_Value']
print("email_group:   ")
print(email_group.describe())
print()
print("social_group:   ")
print(social_group.describe())
print()
```

```python
#TWO SAMPLE Z TEST
z_stat,p_value=ztest(email_group, social_group,alternative='two-sided')
alpha=0.05
if p_value<alpha:
    conclusion="Reject the null hypothesis: significant difference avg purchase value"
else:
    conclusion="failed to reject the null hypothesis: no significant difference avg purchase value"
print(f"z_statistic:{z_stat:.2f}")
print(f"p_value:{p_value:.4f}")
print(f"conclusion:{conclusion}")
```

```python
# TWO SAMPLE T TEST

from scipy import stats
t_stat, p_value=stats.ttest_ind(email_group,social_group)
alpha=0.05
if p_value<alpha:
    conclusion="Reject the null hypothesis: significant difference avg purchase value"
else:
    conclusion="failed to reject the null hypothesis: no significant difference avg purchase value"
print(f"T-statistic:{t_stat:.2f}")
print(f"p_value:{p_value:.4f}")
print(f"conclusion:{conclusion}")
    
```
